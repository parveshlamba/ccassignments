```python
import random
from collections import Counter

roll = 1024160141
random.seed(roll)

nums = [random.randint(101, 901) for _ in range(101)]

print("List:")
print(nums)

odd = [i for i in nums if i % 2 != 0]
print("\nOdd Count:", len(odd))

even = [i for i in nums if i % 2 == 0]
print("Even Count:", len(even))

def prime_check(nv):
    if nv < 2:
        return False
    for kk in range(2, int(nv**0.5) + 1):
        if nv % kk == 0:
            return False
    return True

primes = [i for i in nums if prime_check(i)]
print("Prime Count:", len(primes))
print("Prime Numbers:", primes)

c = Counter(nums)
most = c.most_common(1)[0]
print("Most Frequent Number:", most[0])
print("Occurrences:", most[1])
```

```python
digits = [2,1,3,5,2,7,1,2,4,2]

A = {i * 8 for i in digits}
B = {i * 10 for i in digits}

print("A =", A)
print("B =", B)

print("Union =", A.union(B))

print("Intersection =", A.intersection(B))

print("A-B =", A.difference(B))
print("B-A =", B.difference(A))

print("Symmetric Difference =", A.symmetric_difference(B))

print("A subset of B:", A.issubset(B))
print("B superset of A:", B.issuperset(A))

val_rem = int(input("Enter value to remove from A: "))
A.discard(val_rem)
print("Updated A:", A)
```

```python
import pandas as pd

data = {
    "Name": ["parvesh", "Rahul", "Priya", "Karan"],
    "Age": [21, 22, 21, 23],
    "Marks": [92, 79, 96, 68]
}

df = pd.DataFrame(data)

print(df)
print("\nAverage marks:", df["Marks"].mean())
print("\nTopper:")
print(df.loc[df["Marks"].idxmax()])
```

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": ["parvesh", "Rahul", "Priya", "Karan"],
    "Marks": [92, np.nan, 96, 73],
    "Age": [21, 22, np.nan, 23]
})

print(df)

print("\nMissing values:")
print(df.isnull().sum())

df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
df["Age"] = df["Age"].fillna(df["Age"].median())

print("\nAfter filling:")
print(df)
```
